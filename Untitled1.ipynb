```python
# ============================================================
# 单元格1：初始化环境 + 函数定义
# 版本：v1.0_202501
# ============================================================

import warnings
warnings.filterwarnings('ignore')

# 检查依赖
print("🔍 检查依赖包...")
try:
    import librosa
    import numpy as np
    import tensorflow as tf
    from sklearn.model_selection import train_test_split
    from sklearn.preprocessing import StandardScaler
    from sklearn.utils.class_weight import compute_class_weight
    import matplotlib.pyplot as plt
    import joblib
    import os
    from collections import Counter, defaultdict
    from datetime import datetime
    import json
    print("✅ 所有依赖包已正确导入")
except ImportError as e:
    print(f"❌ 缺少依赖包: {e}")
    print("请运行: pip install librosa numpy tensorflow scikit-learn matplotlib joblib")

# ---------------------------------------------------------
# 1. 读取标签函数（v1.0）
# ---------------------------------------------------------
def read_numeric_labels_v1(label_file_path):
    """读取Audacity导出的标签文件，只读取1和2，自动转为0和1"""
    annotations = []
    label_stats = {'1': 0, '2': 0, 'other': 0}
    
    try:
        with open(label_file_path, 'r', encoding='utf-8') as f:
            for line_num, line in enumerate(f, 1):
                line = line.strip()
                if not line:
                    continue
                    
                parts = line.split('\t')
                if len(parts) < 3:
                    continue
                
                try:
                    start_time = float(parts[0])
                    end_time = float(parts[1])
                    label_str = parts[2].strip()
                    
                    if label_str == '1':
                        annotations.append((start_time, end_time, 0))
                        label_stats['1'] += 1
                    elif label_str == '2':
                        annotations.append((start_time, end_time, 1))
                        label_stats['2'] += 1
                    else:
                        label_stats['other'] += 1
                        
                except ValueError:
                    continue
                    
    except FileNotFoundError:
        print(f"❌ 找不到文件: {label_file_path}")
        return [], label_stats
    
    return annotations, label_stats

# ---------------------------------------------------------
# 2. 特征提取函数（v1.0）
# ---------------------------------------------------------
def extract_enhanced_features_v1(audio_segment, sr):
    """提取音频特征，返回特征数组"""
    features = []
    
    # 1. 时域特征
    rms = librosa.feature.rms(y=audio_segment)[0]
    features.extend([np.mean(rms), np.std(rms), np.max(rms)])
    
    # 2. 过零率
    zcr = librosa.feature.zero_crossing_rate(audio_segment)[0]
    features.extend([np.mean(zcr), np.std(zcr)])
    
    # 3. 频谱质心
    spectral_centroid = librosa.feature.spectral_centroid(y=audio_segment, sr=sr)[0]
    features.extend([np.mean(spectral_centroid), np.std(spectral_centroid)])
    
    # 4. 频谱带宽
    spectral_bandwidth = librosa.feature.spectral_bandwidth(y=audio_segment, sr=sr)[0]
    features.extend([np.mean(spectral_bandwidth), np.std(spectral_bandwidth)])
    
    # 5. MFCC特征（13个系数）
    mfcc = librosa.feature.mfcc(y=audio_segment, sr=sr, n_mfcc=13)
    for i in range(13):
        features.append(np.mean(mfcc[i]))
    
    # 6. 频谱衰减点
    spectral_rolloff = librosa.feature.spectral_rolloff(y=audio_segment, sr=sr)[0]
    features.append(np.mean(spectral_rolloff))
    
    return np.array(features)

# ---------------------------------------------------------
# 3. 滑动检测器函数（v1.0）
# ---------------------------------------------------------
def sliding_detector_v1(y_audio, sr, short_window=1.0, overlap=0.5, context_window=2.0):
    """
    滑动检测器 + 上下文分析
    short_window: 短窗口大小（秒）
    overlap: 短窗口重叠比例
    context_window: 上下文窗口大小（秒）
    """
    # 计算短窗口参数
    short_samples = int(short_window * sr)
    step_samples = int(short_samples * (1 - overlap))
    context_samples = int(context_window * sr)
    
    # 存储检测结果
    detection_points = []
    context_segments = []
    
    # 滑动短窗口
    for start in range(0, len(y_audio) - short_samples, step_samples):
        end = start + short_samples
        short_segment = y_audio[start:end]
        
        # 快速检测特征（简化版）
        rms = np.mean(librosa.feature.rms(y=short_segment)[0])
        zcr = np.mean(librosa.feature.zero_crossing_rate(short_segment)[0])
        
        # 检测阈值（可根据实际数据调整）
        if rms > 0.01 and zcr < 0.3:  # 假设鼾声RMS较大，过零率较低
            detection_time = (start + end) / (2 * sr)  # 检测点的时间
            
            # 提取上下文窗口
            context_start = max(0, start - context_samples)
            context_end = min(len(y_audio), end + context_samples)
            context_segment = y_audio[context_start:context_end]
            
            detection_points.append({
                'time': detection_time,
                'context_start': context_start / sr,
                'context_end': context_end / sr,
                'confidence': rms  # 用RMS作为置信度初步估计
            })
            
            # 保存上下文片段用于特征提取
            context_segments.append({
                'segment': context_segment,
                'start_time': context_start / sr,
                'end_time': context_end / sr,
                'center_time': detection_time
            })
    
    return detection_points, context_segments

# ---------------------------------------------------------
# 4. 合并相邻事件函数（v1.0）
# ---------------------------------------------------------
def merge_events_v1(events, max_gap=0.5):
    """
    合并相邻的相同类型事件
    events: 事件列表，每个事件是字典，包含start_time, end_time, type等
    max_gap: 最大间隔（秒），超过此间隔不合并
    """
    if not events:
        return []
    
    # 按开始时间排序
    events.sort(key=lambda x: x['start_time'])
    
    merged_events = []
    current_event = events[0].copy()
    
    for next_event in events[1:]:
        # 如果类型相同且时间间隔小于阈值，合并
        if (next_event['type'] == current_event['type'] and 
            next_event['start_time'] - current_event['end_time'] <= max_gap):
            current_event['end_time'] = next_event['end_time']
            current_event['duration'] = current_event['end_time'] - current_event['start_time']
            # 合并置信度（取最大值）
            current_event['confidence'] = max(current_event['confidence'], next_event['confidence'])
        else:
            merged_events.append(current_event)
            current_event = next_event.copy()
    
    merged_events.append(current_event)
    return merged_events

# ---------------------------------------------------------
# 5. 用户视角输出函数（v1.0）
# ---------------------------------------------------------
def user_perspective_output_v1(events, total_duration):
    """
    生成用户视角的输出
    events: 合并后的事件列表
    total_duration: 音频总时长（秒）
    """
    print("\n" + "="*80)
    print("👤 用户视角报告")
    print("="*80)
    
    if not events:
        print("📊 未检测到鼾声事件")
        return
    
    type_names = {0: "正常鼾声", 1: "严重鼾声"}
    
    # 统计信息
    type_counts = defaultdict(int)
    total_snore_time = 0
    
    print(f"{'开始时间':<12} {'结束时间':<12} {'持续时间':<10} {'类型':<12} {'时间占比':<10} {'置信度':<8}")
    print("-"*70)
    
    for i, event in enumerate(events, 1):
        duration = event['end_time'] - event['start_time']
        time_ratio = (duration / total_duration) * 100
        
        # 格式化时间输出
        start_str = f"{event['start_time']:.1f}s"
        end_str = f"{event['end_time']:.1f}s"
        duration_str = f"{duration:.1f}s"
        type_str = type_names[event['type']]
        ratio_str = f"{time_ratio:.1f}%"
        conf_str = f"{event['confidence']:.2f}"
        
        print(f"{start_str:<12} {end_str:<12} {duration_str:<10} {type_str:<12} {ratio_str:<10} {conf_str:<8}")
        
        # 统计
        type_counts[event['type']] += 1
        total_snore_time += duration
    
    # 汇总统计
    print("\n" + "="*80)
    print("📈 汇总统计")
    print("-"*80)
    
    print(f"🔹 音频总时长: {total_duration:.1f}秒")
    print(f"🔹 检测到鼾声事件总数: {len(events)}个")
    
    for type_id, type_name in type_names.items():
        count = type_counts[type_id]
        if count > 0:
            print(f"  - {type_name}: {count}个")
    
    print(f"🔹 鼾声总持续时间: {total_snore_time:.1f}秒")
    print(f"🔹 鼾声时间占比: {(total_snore_time/total_duration)*100:.1f}%")
    
    # 保存到文件
    output_data = {
        'analysis_time': datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
        'total_duration': total_duration,
        'total_events': len(events),
        'events': events,
        'summary': {
            'normal_count': type_counts[0],
            'severe_count': type_counts[1],
            'total_snore_time': total_snore_time,
            'snore_ratio': (total_snore_time/total_duration)*100
        }
    }
    
    with open('user_report_v1.json', 'w', encoding='utf-8') as f:
        json.dump(output_data, f, ensure_ascii=False, indent=2)
    
    print(f"✅ 用户报告已保存到: user_report_v1.json")

# ---------------------------------------------------------
# 6. 保存数据到文件（v1.0）
# ---------------------------------------------------------
def save_data_v1(X, y, filename_prefix):
    """保存数据到文件（内存+文件双保险）"""
    # 保存到内存（返回）
    data_dict = {
        'X': X,
        'y': y,
        'timestamp': datetime.now().strftime("%Y%m%d_%H%M%S"),
        'samples': len(X)
    }
    
    # 保存到文件
    filename = f"{filename_prefix}_{data_dict['timestamp']}.npz"
    np.savez(filename, X=X, y=y)
    print(f"💾 数据已保存到文件: {filename}")
    
    return data_dict

print("\n✅ 单元格1：函数定义完成")
print("可用的函数:")
print("  1. read_numeric_labels_v1() - 读取标签")
print("  2. extract_enhanced_features_v1() - 特征提取")
print("  3. sliding_detector_v1() - 滑动检测器")
print("  4. merge_events_v1() - 合并事件")
print("  5. user_perspective_output_v1() - 用户视角输出")
print("  6. save_data_v1() - 保存数据")
```

```python
# ============================================================
# 单元格2：参数对比实验
# 版本：v2.0_202501
# ============================================================

print("🔬 参数对比实验开始")
print("测试窗口配置: 1s(50%重叠)、2s、2s(50%重叠)、5s")

# 检查依赖
if 'read_numeric_labels_v1' not in globals():
    print("❌ 请先运行单元格1")
else:
    print("✅ 依赖检查通过")

# ---------------------------------------------------------
# 2.1 数据探索：分析标签特性
# ---------------------------------------------------------
print("\n" + "="*80)
print("2.1 数据探索：分析标签特性")
print("="*80)

# 文件路径
audio_label_pairs_v2 = [
    ('./my_mp3_snores/snoring-1.mp3', './my_mp3_snores/snoringlable-1.txt'),
    ('./my_mp3_snores/snoring-6.mp3', './my_mp3_snores/snoringlable-6.txt'),
    ('./my_mp3_snores/snoring-3.mp3', './my_mp3_snores/snoringlable-3.txt'),
    ('my_mp3_snores/snoring-4.mp3', './my_mp3_snores/snoringlable-4.txt'),
    ('./my_mp3_snores/snoring-5.mp3', './my_mp3_snores/snoringlable-5.txt'),
]

def analyze_labels_v2(pairs):
    """分析标签文件的特性"""
    all_annotations = []
    label_durations = {'normal': [], 'severe': []}
    
    for audio_path, label_path in pairs:
        if not os.path.exists(label_path):
            print(f"⚠️  跳过：找不到标签 {label_path}")
            continue
            
        annotations, stats = read_numeric_labels_v1(label_path)
        print(f"📄 {os.path.basename(label_path)}: 正常{stats['1']}个, 严重{stats['2']}个")
        
        for ann in annotations:
            all_annotations.append(ann)
            duration = ann[1] - ann[0]
            if ann[2] == 0:
                label_durations['normal'].append(duration)
            else:
                label_durations['severe'].append(duration)
    
    # 分析持续时间
    print("\n📊 标签持续时间分析:")
    for label_type, durations in label_durations.items():
        if durations:
            print(f"  {label_type}: {len(durations)}个, 平均{np.mean(durations):.1f}秒, "
                  f"最短{np.min(durations):.1f}秒, 最长{np.max(durations):.1f}秒")
    
    return label_durations

label_durations_v2 = analyze_labels_v2(audio_label_pairs_v2)

# ---------------------------------------------------------
# 2.2 快速筛选：测试多种参数
# ---------------------------------------------------------
print("\n" + "="*80)
print("2.2 快速筛选：测试多种参数")
print("="*80)

def create_dataset_v2(audio_path, label_path, segment_length=2.0, overlap=0.0):
    """创建数据集 - 优化版"""
    print(f"  🔍 处理: {os.path.basename(audio_path)}")
    
    # 读取标签
    annotations, label_stats = read_numeric_labels_v1(label_path)
    if len(annotations) == 0:
        return None, None
    
    # 加载音频
    y_audio, sr = librosa.load(audio_path, sr=None)
    total_duration = len(y_audio) / sr
    
    # 参数计算
    segment_samples = int(segment_length * sr)
    step_samples = int(segment_samples * (1 - overlap))
    
    X_segments = []
    y_segments = []
    confidence_scores = []
    
    # 滑动窗口
    for start in range(0, len(y_audio) - segment_samples, step_samples):
        end = start + segment_samples
        seg_start_time = start / sr
        seg_end_time = end / sr
        
        # 统计标签时间
        time_0 = 0.0  # 正常鼾声
        time_1 = 0.0  # 严重鼾声
        
        for ann_start, ann_end, label in annotations:
            overlap_start = max(ann_start, seg_start_time)
            overlap_end = min(ann_end, seg_end_time)
            overlap_duration = max(0, overlap_end - overlap_start)
            
            if overlap_duration > 0:
                if label == 0:
                    time_0 += overlap_duration
                else:
                    time_1 += overlap_duration
        
        total_label_time = time_0 + time_1
        
        # 确定标签
        if total_label_time > 0:
            coverage_ratio = total_label_time / segment_length
            
            if coverage_ratio > 0.3:  # 至少30%的覆盖率
                if time_0 >= time_1:
                    label = 0
                    confidence = time_0 / total_label_time
                else:
                    label = 1
                    confidence = time_1 / total_label_time
                
                if confidence > 0.6:  # 置信度阈值提高到0.6
                    segment = y_audio[start:end]
                    feat = extract_enhanced_features_v1(segment, sr)
                    X_segments.append(feat)
                    y_segments.append(label)
                    confidence_scores.append(confidence)
    
    if len(X_segments) == 0:
        return None, None
    
    X = np.array(X_segments)
    y = np.array(y_segments)
    
    class_counts = Counter(y)
    print(f"    ✅ 生成 {len(X)} 个片段 (正常:{class_counts.get(0,0)}, "
          f"严重:{class_counts.get(1,0)}, 平均置信度:{np.mean(confidence_scores):.2f})")
    
    return X, y

def process_multiple_v2(pairs, segment_length=2.0, overlap=0.0, config_name=""):
    """处理多个文件 - 优化版"""
    print(f"\n⚙️  配置: {config_name} (窗口:{segment_length}s, 重叠:{overlap})")
    
    all_X, all_y = [], []
    
    for audio_path, label_path in pairs:
        if not os.path.exists(audio_path):
            print(f"⚠️  跳过：找不到音频 {audio_path}")
            continue
        if not os.path.exists(label_path):
            print(f"⚠️  跳过：找不到标签 {label_path}")
            continue
            
        X, y = create_dataset_v2(audio_path, label_path, segment_length, overlap)
        if X is not None:
            all_X.append(X)
            all_y.append(y)
    
    if not all_X:
        return None, None
    
    X_combined = np.vstack(all_X)
    y_combined = np.concatenate(all_y)
    
    # 保存数据
    data_dict = save_data_v1(X_combined, y_combined, f"config_{config_name}")
    
    return X_combined, y_combined

# 测试不同配置
configs_v2 = [
    {'segment_length': 1.0, 'overlap': 0.5, 'name': '1s_50%'},
    {'segment_length': 2.0, 'overlap': 0.0, 'name': '2s_0%'},
    {'segment_length': 2.0, 'overlap': 0.5, 'name': '2s_50%'},
    {'segment_length': 5.0, 'overlap': 0.0, 'name': '5s_0%'},
]

results_v2 = []

for config in configs_v2:
    X, y = process_multiple_v2(
        audio_label_pairs_v2,
        segment_length=config['segment_length'],
        overlap=config['overlap'],
        config_name=config['name']
    )
    
    if X is not None:
        results_v2.append({
            'name': config['name'],
            'X': X,
            'y': y,
            'segment_length': config['segment_length'],
            'overlap': config['overlap'],
            'samples': len(X),
            'normal_count': np.sum(y == 0),
            'severe_count': np.sum(y == 1)
        })

# ---------------------------------------------------------
# 2.3 精细验证：小规模模型验证
# ---------------------------------------------------------
print("\n" + "="*80)
print("2.3 精细验证：小规模模型验证")
print("="*80)

def quick_train_eval_v2(X, y, config_name, epochs=15):
    """快速训练和评估"""
    print(f"\n🧪 快速验证配置: {config_name}")
    
    # 标准化
    scaler_v2 = StandardScaler()
    X_scaled = scaler_v2.fit_transform(X)
    
    # 划分数据集
    X_train, X_test, y_train, y_test = train_test_split(
        X_scaled, y, test_size=0.2, random_state=42, stratify=y
    )
    
    # 构建简单模型
    model_v2 = tf.keras.Sequential([
        tf.keras.layers.Dense(32, activation='relu', input_shape=(X_train.shape[1],)),
        tf.keras.layers.Dropout(0.2),
        tf.keras.layers.Dense(16, activation='relu'),
        tf.keras.layers.Dense(2, activation='softmax')
    ])
    
    model_v2.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )
    
    # 训练
    history_v2 = model_v2.fit(
        X_train, y_train,
        epochs=epochs,
        batch_size=16,
        validation_split=0.2,
        verbose=0
    )
    
    # 评估
    loss, acc = model_v2.evaluate(X_test, y_test, verbose=0)
    
    print(f"   📊 测试准确率: {acc:.2%}")
    print(f"   📊 最终验证准确率: {history_v2.history['val_accuracy'][-1]:.2%}")
    
    return acc, history_v2

# 对每个配置进行快速验证
for result in results_v2:
    acc, history = quick_train_eval_v2(
        result['X'], result['y'], result['name']
    )
    result['accuracy'] = acc

# ---------------------------------------------------------
# 2.4 确定参数：输出最佳配置
# ---------------------------------------------------------
print("\n" + "="*80)
print("2.4 确定参数：输出最佳配置")
print("="*80)

# 显示对比结果
print("\n📊 参数对比结果:")
print("-"*80)
print("配置\t\t总样本\t正常\t严重\t准确率")
print("-"*80)

for result in results_v2:
    print(f"{result['name']:<10}\t{result['samples']:<6}\t"
          f"{result['normal_count']:<4}\t{result['severe_count']:<4}\t"
          f"{result['accuracy']:.2%}")

# 选择最佳配置（基于准确率和样本数量）
best_config_v2 = max(results_v2, key=lambda x: (x['accuracy'], x['samples']))

print(f"\n🎯 最佳配置: {best_config_v2['name']}")
print(f"   - 窗口大小: {best_config_v2['segment_length']}秒")
print(f"   - 重叠比例: {best_config_v2['overlap']}")
print(f"   - 总样本数: {best_config_v2['samples']}个")
print(f"   - 测试准确率: {best_config_v2['accuracy']:.2%}")

# 保存最佳配置
best_config_info = {
    'name': best_config_v2['name'],
    'segment_length': best_config_v2['segment_length'],
    'overlap': best_config_v2['overlap'],
    'accuracy': float(best_config_v2['accuracy']),
    'selected_time': datetime.now().strftime("%Y-%m-%d %H:%M:%S")
}

with open('best_config_v2.json', 'w', encoding='utf-8') as f:
    json.dump(best_config_info, f, ensure_ascii=False, indent=2)

print(f"✅ 最佳配置已保存到: best_config_v2.json")
print("🎉 参数对比实验完成！")
```

```python
# ============================================================
# 单元格4：训练最终模型
# 版本：v4.0_202501
# ============================================================

print("🧠 训练最终模型")

# 检查依赖和数据
if 'X_all_v3' not in globals() or 'y_all_v3' not in globals():
    print("❌ 请先运行单元格3获取数据")
else:
    print(f"✅ 数据准备就绪: {len(X_all_v3)}个样本")

def train_final_model_v4(X_all, y_all, model_name="final_snore_model"):
    """训练最终模型 - 增强版"""
    print(f"\n🚀 开始训练最终模型...")
    print(f"   数据形状: X{X_all.shape}, y{y_all.shape}")
    print(f"   类别分布: 正常{np.sum(y_all==0)}个, 严重{np.sum(y_all==1)}个")
    
    # 1. 标准化
    scaler_v4 = StandardScaler()
    X_scaled_v4 = scaler_v4.fit_transform(X_all)
    
    # 2. 划分数据集
    X_train_v4, X_test_v4, y_train_v4, y_test_v4 = train_test_split(
        X_scaled_v4, y_all, 
        test_size=0.2, 
        random_state=42,
        stratify=y_all
    )
    
    print(f"   训练集: {len(y_train_v4)}个 (正常:{np.sum(y_train_v4==0)}, 严重:{np.sum(y_train_v4==1)})")
    print(f"   测试集: {len(y_test_v4)}个 (正常:{np.sum(y_test_v4==0)}, 严重:{np.sum(y_test_v4==1)})")
    
    # 3. 构建优化模型
    model_v4 = tf.keras.Sequential([
        tf.keras.layers.Dense(64, activation='relu', input_shape=(X_train_v4.shape[1],),
                             kernel_regularizer=tf.keras.regularizers.l2(0.001)),
        tf.keras.layers.Dropout(0.3),
        tf.keras.layers.Dense(32, activation='relu',
                             kernel_regularizer=tf.keras.regularizers.l2(0.001)),
        tf.keras.layers.Dropout(0.3),
        tf.keras.layers.Dense(16, activation='relu'),
        tf.keras.layers.Dense(2, activation='softmax')
    ])
    
    # 4. 编译模型
    model_v4.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=0.0005),
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy', 
                tf.keras.metrics.Precision(name='precision'),
                tf.keras.metrics.Recall(name='recall')]
    )
    
    # 5. 添加早停和模型检查点
    callbacks_v4 = [
        tf.keras.callbacks.EarlyStopping(
            monitor='val_loss',
            patience=10,
            restore_best_weights=True
        ),
        tf.keras.callbacks.ModelCheckpoint(
            filepath=f'{model_name}_best.keras',
            monitor='val_accuracy',
            save_best_only=True
        ),
        tf.keras.callbacks.ReduceLROnPlateau(
            monitor='val_loss',
            factor=0.5,
            patience=5,
            min_lr=0.00001
        )
    ]
    
    # 6. 训练
    print("\n⏳ 训练中...")
    history_v4 = model_v4.fit(
        X_train_v4, y_train_v4,
        epochs=50,
        batch_size=16,
        validation_split=0.2,
        callbacks=callbacks_v4,
        verbose=1
    )
    
    # 7. 加载最佳模型
    model_v4 = tf.keras.models.load_model(f'{model_name}_best.keras')
    
    # 8. 评估
    print("\n📊 模型评估:")
    results_v4 = model_v4.evaluate(X_test_v4, y_test_v4, verbose=0)
    
    metrics_names = ['损失', '准确率', '精确率', '召回率']
    for name, value in zip(metrics_names, results_v4):
        print(f"   {name}: {value:.4f}")
    
    # 9. 预测并显示详细报告
    y_pred_v4 = model_v4.predict(X_test_v4, verbose=0)
    y_pred_classes_v4 = np.argmax(y_pred_v4, axis=1)
    
    from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score
    
    print("\n📋 详细分类报告:")
    print(classification_report(y_test_v4, y_pred_classes_v4, 
                               target_names=['正常鼾声', '严重鼾声'],
                               digits=4))
    
    # 计算AUC
    try:
        auc_score = roc_auc_score(y_test_v4, y_pred_v4[:, 1])
        print(f"📈 AUC分数: {auc_score:.4f}")
    except:
        print("⚠️  AUC计算失败（可能类别不平衡）")
    
    # 10. 绘制训练曲线
    plt.figure(figsize=(15, 5))
    
    # 准确率曲线
    plt.subplot(1, 3, 1)
    plt.plot(history_v4.history['accuracy'], label='训练准确率')
    plt.plot(history_v4.history['val_accuracy'], label='验证准确率')
    plt.title('模型准确率')
    plt.xlabel('Epoch')
    plt.ylabel('准确率')
    plt.legend()
    plt.grid(True)
    
    # 损失曲线
    plt.subplot(1, 3, 2)
    plt.plot(history_v4.history['loss'], label='训练损失')
    plt.plot(history_v4.history['val_loss'], label='验证损失')
    plt.title('模型损失')
    plt.xlabel('Epoch')
    plt.ylabel('损失')
    plt.legend()
    plt.grid(True)
    
    # 混淆矩阵
    plt.subplot(1, 3, 3)
    cm = confusion_matrix(y_test_v4, y_pred_classes_v4)
    plt.imshow(cm, interpolation='nearest', cmap=plt.cm.Blues)
    plt.title('混淆矩阵')
    plt.colorbar()
    
    # 添加数字标签
    thresh = cm.max() / 2.
    for i in range(cm.shape[0]):
        for j in range(cm.shape[1]):
            plt.text(j, i, format(cm[i, j], 'd'),
                    ha="center", va="center",
                    color="white" if cm[i, j] > thresh else "black")
    
    plt.xticks([0, 1], ['正常', '严重'])
    plt.yticks([0, 1], ['正常', '严重'])
    plt.ylabel('真实标签')
    plt.xlabel('预测标签')
    
    plt.tight_layout()
    plt.show()
    
    # 11. 保存最终模型和Scaler
    model_v4.save(f'{model_name}_final.keras')
    joblib.dump(scaler_v4, f'{model_name}_scaler.pkl')
    
    print(f"\n✅ 模型训练完成!")
    print(f"   💾 模型已保存: {model_name}_final.keras")
    print(f"   💾 Scaler已保存: {model_name}_scaler.pkl")
    
    # 12. 返回模型信息
    model_info_v4 = {
        'model': model_v4,
        'scaler': scaler_v4,
        'history': history_v4,
        'test_results': dict(zip(['loss', 'accuracy', 'precision', 'recall'], results_v4)),
        'config': best_config_v2
    }
    
    return model_info_v4

# 训练最终模型
if 'X_all_v3' in globals() and 'y_all_v3' in globals():
    model_info_v4 = train_final_model_v4(X_all_v3, y_all_v3, "snore_detector_v4")
    print("\n🎉 单元格4完成！")
```

```python
# ============================================================
# 单元格5：预测推理
# 版本：v5.0_202501
# ============================================================

print("🎯 预测推理：滑动检测 + 用户视角输出")

# 检查依赖
if 'model_info_v4' not in globals():
    print("❌ 请先运行单元格4训练模型")
    print("⚠️  使用示例模型进行演示...")
    # 创建示例模型（仅用于演示）
    model_v5 = None
    scaler_v5 = None
else:
    model_v5 = model_info_v4['model']
    scaler_v5 = model_info_v4['scaler']
    print(f"✅ 模型加载成功: 准确率{model_info_v4['test_results']['accuracy']:.2%}")

# ---------------------------------------------------------
# 完整的预测流程
# ---------------------------------------------------------
def predict_audio_v5(audio_path, model, scaler, 
                    short_window=1.0, overlap=0.5, 
                    context_window=2.0, merge_gap=0.5):
    """
    完整的预测流程
    1. 加载音频
    2. 滑动检测找到候选点
    3. 上下文分析提取特征
    4. 模型预测
    5. 合并相邻事件
    6. 用户视角输出
    """
    print(f"\n🎧 分析音频: {os.path.basename(audio_path)}")
    
    # 1. 加载音频
    y_audio, sr = librosa.load(audio_path, sr=None)
    total_duration = len(y_audio) / sr
    print(f"   音频信息: {total_duration:.1f}秒, 采样率: {sr}Hz")
    
    # 2. 滑动检测器找到候选点
    print("   步骤1: 滑动检测器中...")
    detection_points, context_segments = sliding_detector_v1(
        y_audio, sr, 
        short_window=short_window, 
        overlap=overlap,
        context_window=context_window
    )
    
    print(f"   检测到 {len(detection_points)} 个候选点")
    
    if len(detection_points) == 0:
        print("⚠️  未检测到候选点，可能是静音或阈值设置不当")
        user_perspective_output_v1([], total_duration)
        return []
    
    # 3. 提取特征并预测
    print("   步骤2: 提取特征和预测中...")
    events_v5 = []
    
    for i, seg_info in enumerate(context_segments):
        # 提取特征
        features = extract_enhanced_features_v1(seg_info['segment'], sr)
        
        if scaler is not None:
            # 标准化特征
            features_scaled = scaler.transform(features.reshape(1, -1))
        else:
            features_scaled = features.reshape(1, -1)
        
        # 预测
        if model is not None:
            prediction = model.predict(features_scaled, verbose=0)
            predicted_class = np.argmax(prediction[0])
            confidence = prediction[0][predicted_class]
        else:
            # 示例预测（仅用于演示）
            predicted_class = 0 if i % 3 == 0 else 1
            confidence = 0.7 + np.random.random() * 0.3
        
        # 创建事件
        event_v5 = {
            'start_time': seg_info['start_time'],
            'end_time': seg_info['end_time'],
            'center_time': seg_info['center_time'],
            'type': predicted_class,
            'confidence': float(confidence),
            'duration': seg_info['end_time'] - seg_info['start_time']
        }
        
        events_v5.append(event_v5)
    
    # 4. 合并相邻事件
    print("   步骤3: 合并相邻事件中...")
    
    # 首先按中心时间排序
    events_v5.sort(key=lambda x: x['center_time'])
    
    # 合并相邻的相同类型事件
    merged_events_v5 = merge_events_v1(events_v5, max_gap=merge_gap)
    
    print(f"   合并后事件: {len(merged_events_v5)} 个")
    
    # 5. 用户视角输出
    print("   步骤4: 生成用户报告...")
    user_perspective_output_v1(merged_events_v5, total_duration)
    
    # 6. 可视化结果
    visualize_results_v5(y_audio, sr, merged_events_v5)
    
    return merged_events_v5

def visualize_results_v5(y_audio, sr, events):
    """可视化结果"""
    plt.figure(figsize=(15, 10))
    
    # 1. 原始波形
    time_axis = np.linspace(0, len(y_audio)/sr, len(y_audio))
    
    plt.subplot(3, 1, 1)
    plt.plot(time_axis, y_audio, color='blue', alpha=0.6, linewidth=0.5)
    plt.title('原始音频波形')
    plt.xlabel('时间 (秒)')
    plt.ylabel('振幅')
    plt.grid(True, alpha=0.3)
    
    # 2. 频谱图
    plt.subplot(3, 1, 2)
    D = librosa.amplitude_to_db(np.abs(librosa.stft(y_audio)), ref=np.max)
    librosa.display.specshow(D, sr=sr, x_axis='time', y_axis='log')
    plt.colorbar(format='%+2.0f dB')
    plt.title('频谱图')
    
    # 3. 事件标记
    plt.subplot(3, 1, 3)
    
    # 绘制时间轴
    plt.plot([0, time_axis[-1]], [0, 0], 'k-', alpha=0.3)
    
    # 用不同颜色标记事件
    colors = {0: 'green', 1: 'red'}
    labels = {0: '正常鼾声', 1: '严重鼾声'}
    
    for event in events:
        color = colors[event['type']]
        label = labels[event['type']]
        
        # 绘制事件区间
        plt.fill_between(
            [event['start_time'], event['end_time']],
            -0.5, 0.5,
            alpha=0.3,
            color=color,
            label=label if event == events[0] else ""
        )
        
        # 标注事件类型
        center = (event['start_time'] + event['end_time']) / 2
        plt.text(center, 0.3, label, 
                ha='center', va='center',
                fontsize=9, fontweight='bold',
                bbox=dict(boxstyle="round,pad=0.3", 
                         facecolor='white', alpha=0.8))
    
    plt.title('鼾声事件检测结果')
    plt.xlabel('时间 (秒)')
    plt.ylabel('事件类型')
    plt.ylim(-1, 1)
    plt.yticks([])
    plt.grid(True, alpha=0.3)
    plt.legend(loc='upper right')
    
    plt.tight_layout()
    plt.show()

# ---------------------------------------------------------
# 运行预测
# ---------------------------------------------------------
# 选择一个测试音频
test_audio_v5 = './my_mp3_snores/snoring-1.mp3'

if os.path.exists(test_audio_v5):
    # 执行预测
    events_result_v5 = predict_audio_v5(
        test_audio_v5,
        model_v5,
        scaler_v5,
        short_window=1.0,      # 1秒短窗口
        overlap=0.5,          # 50%重叠
        context_window=2.0,   # 前后各2秒上下文
        merge_gap=0.5         # 0.5秒内合并
    )
    
    # 保存预测结果
    if events_result_v5:
        result_data_v5 = {
            'audio_file': test_audio_v5,
            'analysis_time': datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
            'parameters': {
                'short_window': 1.0,
                'overlap': 0.5,
                'context_window': 2.0,
                'merge_gap': 0.5
            },
            'events': events_result_v5
        }
        
        with open('prediction_results_v5.json', 'w', encoding='utf-8') as f:
            json.dump(result_data_v5, f, ensure_ascii=False, indent=2)
        
        print(f"\n✅ 预测结果已保存到: prediction_results_v5.json")
else:
    print(f"⚠️  测试音频不存在: {test_audio_v5}")
    print("请修改 test_audio_v5 变量为有效的音频路径")

print("\n" + "="*80)
print("📋 代码结构总结")
print("="*80)
print("1️⃣ 单元格1: 初始化环境 + 函数定义")
print("   - 基础函数、滑动检测器、用户视角输出")
print("")
print("2️⃣ 单元格2: 参数对比实验")
print("   - 测试4种窗口配置，选择最佳参数")
print("   - 快速模型验证，确定最佳配置")
print("")
print("3️⃣ 单元格3: 用最佳参数处理全部数据")
print("   - 使用最佳配置处理所有训练数据")
print("   - 保存最终数据集")
print("")
print("4️⃣ 单元格4: 训练最终模型")
print("   - 训练优化模型，添加早停和检查点")
print("   - 保存模型和scaler")
print("")
print("5️⃣ 单元格5: 预测推理")
print("   - 滑动检测 + 上下文分析")
print("   - 用户视角报告输出")
print("   - 结果可视化")
print("")
print("🎉 全部代码完成！")
```
